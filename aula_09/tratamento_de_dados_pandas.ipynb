### **2.3 Tratamento da coluna idade**

```python
# Números padronizados
numeros = {
    "vinte": 20,
    "vinte e um": 21,
    "vinte e dois": 22,
    "vinte e três": 23,
    "vinte e quatro": 24,
    "vinte e cinco": 25,
    "vinte e seis": 26,
    "vinte e sete": 27,
    "vinte e oito": 28,
    "vinte e nove": 29,
    "trinta": 30,
}


def padronizar_idade(valor):

    # 1. Valores nulos ou vazios
    if pd.isna(valor):
        return "Não informado"

    valor = str(valor).lower().strip()

    if valor == "":
        return "Não informado"

    # 2. Remove "anos"
    valor = valor.replace("anos", "").strip()

    # 3. Verifica se é número inteiro
    try:
        idade = int(valor)
    except ValueError:

        # 4. Procura número por extenso
        if valor in numeros:
            idade = numeros[valor]
        else:
            return "Não informado"

    # 5. Verifica se a idade está dentro de um intervalo válido
    if idade < 0 or idade > 80:
        return "Não informado"

    return idade
```

```python
# Aplicando a função de tratamento da coluna idade
df["idade"] = df["idade"].apply(padronizar_idade)
```

### **2.5 Tratamento da coluna email**

```python
def padronizar_email(email):

    # Valores ausentes
    if pd.isna(email) or str(email).strip() == "":
        return "Não informado"

    # Converte para texto, remove espaços e coloca em minúsculo
    email = str(email).strip().lower().replace(" ", "")

    # Se houver mais de um @, mantém apenas o primeiro
    if email.count("@") > 1:
        usuario = email.split("@")[0]
        return usuario + "@email.com"

    # Se não existir @
    if "@" not in email:
        # Tenta separar "rafaelemail.com" → "rafael"
        if "email.com" in email:
            usuario = email.split("email.com")[0]
            return usuario + "@email.com"

        # Caso contrário, considera inválido
        return "Email inválido"

    # Pega somente a parte antes do @
    usuario = email.split("@")[0]

    if usuario == "":
        return "Email inválido"

    # Padroniza o domínio
    return usuario + "@email.com"
```

```python
# Aplicando a função de tratamento de email
df["email"] = df["email"].apply(padronizar_email)
df
```

### **2.7 Tratamento da coluna CPF**

```python
def padronizar_cpf(cpf):

    # Valores ausentes ou vazios
    if pd.isna(cpf) or str(cpf).strip() == "":
        return "Não informado"

    # Converte para texto e remove espaços
    cpf = str(cpf).strip()

    # Remove pontos e hífen
    cpf = cpf.replace(".", "").replace("-", "").replace(" ", "")

    # Verifica se possui somente números
    if not cpf.isdigit():
        return "CPF inválido"

    # Verifica se possui exatamente 11 números
    if len(cpf) != 11:
        return "CPF inválido"

    # Formata o CPF: 000.000.000-00
    return f"{cpf[:3]}.{cpf[3:6]}.{cpf[6:9]}-{cpf[9:]}"

# Aplicando a função de tratamento do CPF
df["cpf"] = df["cpf"].apply(padronizar_cpf)

# Verificando os valores da coluna CPF
df["cpf"]
```

### **2.9 valor_inscricao**

```python
def padronizar_valor(valor):
    # 1. Se for nulo ou vazio, assume o valor padrão de 50.0
    if pd.isna(valor) or str(valor).strip() == "":
        return 50.0

    valor_str = str(valor).lower().strip()

    # 2. Corrige o texto por extenso
    if "cinquenta" in valor_str:
        return 50.0

    try:
        num = float(valor_str)
        # 3. Se for negativo (ex: -50.0), transforma em positivo. Se for zero, assume 50.0
        return abs(num) if num != 0 else 50.0
    except ValueError:
        return 50.0

# Aplicando a função na coluna
df["valor_inscricao"] = df["valor_inscricao"].apply(padronizar_valor)

# Exibe o DataFrame final totalmente limpo
df
```
